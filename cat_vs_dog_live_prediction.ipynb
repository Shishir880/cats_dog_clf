```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import img_to_array

# মডেল লোড করো
model = load_model("cats_vs_dogs_model.h5")

# ক্যামেরা অন করো
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # ইনপুট প্রসেস করো
    image = cv2.resize(frame, (150, 150))
    image = image.astype("float32") / 255.0
    image = img_to_array(image)
    image = np.expand_dims(image, axis=0)  # (1, 150, 150, 3)

    # প্রেডিকশন
    prediction = model.predict(image)[0][0]  # e.g., 0.87
    label = ""
    color = (0, 0, 0)

    # থ্রেশহোল্ড অনুসারে ক্লাস নির্ধারণ
    if prediction > 0.75:
        label = f"Dog ({prediction:.2f})"
        color = (0, 255, 0)  # Green
    elif prediction < 0.25:
        label = f"Cat ({prediction:.2f})"
        color = (255, 0, 0)  # Blue
    else:
        label = f"Uncertain ({prediction:.2f})"
        color = (0, 0, 255)  # Red

    # ফ্রেমে টেক্সট দেখাও
    cv2.putText(frame, f"Prediction: {label}", (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 1, color, 2)
    cv2.imshow("Live Cat vs Dog Detection", frame)

    # 'q' চাপলে বের হও
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
